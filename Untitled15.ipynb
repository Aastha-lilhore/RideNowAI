```python
# ============================================================
# RIDENOW AI — COMPLETE ML PIPELINE
# EDA + PREPROCESSING + FEATURE ENGINEERING
# + MODEL TRAINING + EVALUATION + SAVING
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import warnings

from sklearn.model_selection import train_test_split
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    classification_report,
    confusion_matrix
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. LOAD DATA
# ============================================================

FILE_PATH = "/content/RideNow_AI_250k_33col_dataset.csv"

df = pd.read_csv(FILE_PATH)

print("=" * 60)
print("RIDENOW AI — DATASET INFORMATION")
print("=" * 60)

print("Dataset shape:", df.shape)
print("Rows:", len(df))
print("Columns:", len(df.columns))

# ============================================================
# 2. BASIC DATA INSPECTION
# ============================================================

print("\n--- First 5 Rows ---")
display(df.head())

print("\n--- Data Types ---")
print(df.dtypes)

print("\n--- Missing Values ---")
missing = df.isnull().sum()
print(missing[missing > 0] if missing.sum() > 0 else "No missing values")

print("\n--- Duplicate Rows ---")
print("Duplicates:", df.duplicated().sum())

print("\n--- Statistical Summary ---")
display(df.describe().T)

# ============================================================
# 3. TARGET DISTRIBUTION
# ============================================================

print("\n--- Target Distribution ---")

target_counts = df["cancelled"].value_counts()

print(target_counts)
print("\nCancellation rate:",
      round(df["cancelled"].mean() * 100, 2), "%")

plt.figure(figsize=(7, 5))
df["cancelled"].value_counts().sort_index().plot(kind="bar")
plt.title("Ride Cancellation Distribution")
plt.xlabel("Cancelled (0 = Completed, 1 = Cancelled)")
plt.ylabel("Number of Rides")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# ============================================================
# 4. EDA — CANCELLATION BY BOOKING HOUR
# ============================================================

hour_cancel = df.groupby("booking_hour")["cancelled"].mean() * 100

plt.figure(figsize=(10, 5))
hour_cancel.plot(kind="line", marker="o")
plt.title("Cancellation Rate by Booking Hour")
plt.xlabel("Booking Hour")
plt.ylabel("Cancellation Rate (%)")
plt.grid(True)
plt.tight_layout()
plt.show()

# ============================================================
# 5. EDA — TRAFFIC LEVEL
# ============================================================

traffic_cancel = df.groupby("traffic_level")["cancelled"].mean() * 100

plt.figure(figsize=(7, 5))
traffic_cancel.plot(kind="bar")
plt.title("Cancellation Rate by Traffic Level")
plt.xlabel("Traffic Level")
plt.ylabel("Cancellation Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# ============================================================
# 6. EDA — WEATHER
# ============================================================

weather_cancel = df.groupby("weather")["cancelled"].mean() * 100

plt.figure(figsize=(8, 5))
weather_cancel.plot(kind="bar")
plt.title("Cancellation Rate by Weather")
plt.xlabel("Weather")
plt.ylabel("Cancellation Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# ============================================================
# 7. EDA — VEHICLE TYPE
# ============================================================

vehicle_cancel = df.groupby("vehicle_type")["cancelled"].mean() * 100

plt.figure(figsize=(8, 5))
vehicle_cancel.plot(kind="bar")
plt.title("Cancellation Rate by Vehicle Type")
plt.xlabel("Vehicle Type")
plt.ylabel("Cancellation Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# ============================================================
# 8. EDA — PEAK HOURS
# ============================================================

peak_cancel = df.groupby("is_peak_hour")["cancelled"].mean() * 100

plt.figure(figsize=(7, 5))
peak_cancel.plot(kind="bar")
plt.title("Cancellation Rate: Peak vs Non-Peak Hours")
plt.xlabel("Peak Hour (0 = No, 1 = Yes)")
plt.ylabel("Cancellation Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# ============================================================
# 9. EDA — ESTIMATED ARRIVAL TIME
# ============================================================

plt.figure(figsize=(8, 5))
df.groupby("cancelled")["estimated_arrival_min"].mean().plot(
    kind="bar"
)
plt.title("Average Estimated Arrival Time by Cancellation Status")
plt.xlabel("Cancelled (0 = Completed, 1 = Cancelled)")
plt.ylabel("Average ETA (minutes)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# ============================================================
# 10. EDA — DRIVER ACCEPTANCE RATE
# ============================================================

plt.figure(figsize=(8, 5))

df.boxplot(
    column="driver_acceptance_rate",
    by="cancelled"
)

plt.title("Driver Acceptance Rate vs Cancellation")
plt.suptitle("")
plt.xlabel("Cancelled (0 = Completed, 1 = Cancelled)")
plt.ylabel("Driver Acceptance Rate")
plt.tight_layout()
plt.show()

# ============================================================
# 11. EDA — DRIVER CANCELLATION HISTORY
# ============================================================

history_cancel = df.groupby(
    "driver_cancellation_history"
)["cancelled"].mean() * 100

plt.figure(figsize=(10, 5))
history_cancel.plot(kind="line", marker="o")
plt.title("Cancellation Rate vs Driver Cancellation History")
plt.xlabel("Previous Driver Cancellations")
plt.ylabel("Cancellation Rate (%)")
plt.grid(True)
plt.tight_layout()
plt.show()

# ============================================================
# 12. CORRELATION ANALYSIS
# ============================================================

numeric_df = df.select_dtypes(include=np.number)

correlation = numeric_df.corr()["cancelled"].sort_values(
    ascending=False
)

print("\n--- Features Most Related to Cancellation ---")
print(correlation.head(10))

plt.figure(figsize=(10, 8))

top_corr = numeric_df.corr()["cancelled"].sort_values(
    ascending=False
).head(12)

sns.heatmap(
    numeric_df[top_corr.index].corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap — Important Numerical Features")
plt.tight_layout()
plt.show()

# ============================================================
# 13. FEATURE PREPARATION
# ============================================================

print("\n" + "=" * 60)
print("PREPROCESSING")
print("=" * 60)

target = "cancelled"

# Remove target and ID
X = df.drop(columns=[target, "ride_id"]).copy()
y = df[target].astype(int)

# Identify categorical columns
categorical_cols = X.select_dtypes(
    include=["object"]
).columns.tolist()

print("Categorical columns:")
print(categorical_cols)

# Encode categorical variables
for col in categorical_cols:
    X[col] = X[col].astype("category").cat.codes

# Convert to float32 to reduce memory usage
X = X.astype(np.float32)

print("\nFinal feature matrix:", X.shape)
print("Target shape:", y.shape)

# ============================================================
# 14. TRAIN / TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))

# ============================================================
# 15. MODEL TRAINING
# ============================================================

print("\n" + "=" * 60)
print("MODEL TRAINING")
print("=" * 60)

model = HistGradientBoostingClassifier(
    max_iter=120,
    learning_rate=0.08,
    max_leaf_nodes=31,
    min_samples_leaf=80,
    l2_regularization=1.0,
    early_stopping=True,
    random_state=42
)

model.fit(X_train, y_train)

print("✅ Model training completed")

# ============================================================
# 16. MODEL PREDICTIONS
# ============================================================

probabilities = model.predict_proba(X_test)[:, 1]

# Standard threshold
standard_threshold = 0.50

standard_predictions = (
    probabilities >= standard_threshold
).astype(int)

# ============================================================
# 17. STANDARD MODEL EVALUATION
# ============================================================

print("\n" + "=" * 60)
print("MODEL RESULTS — THRESHOLD 0.50")
print("=" * 60)

accuracy = accuracy_score(
    y_test, standard_predictions
)

precision = precision_score(
    y_test, standard_predictions, zero_division=0
)

recall = recall_score(
    y_test, standard_predictions, zero_division=0
)

f1 = f1_score(
    y_test, standard_predictions, zero_division=0
)

roc_auc = roc_auc_score(
    y_test, probabilities
)

pr_auc = average_precision_score(
    y_test, probabilities
)

print(f"Accuracy  : {accuracy*100:.2f}%")
print(f"Precision : {precision*100:.2f}%")
print(f"Recall    : {recall*100:.2f}%")
print(f"F1 Score  : {f1*100:.2f}%")
print(f"ROC-AUC   : {roc_auc:.4f}")
print(f"PR-AUC    : {pr_auc:.4f}")

# ============================================================
# 18. FIND OPTIMAL THRESHOLD
# ============================================================

print("\nFinding optimal decision threshold...")

thresholds = np.arange(0.05, 0.51, 0.01)

best_threshold = 0.50
best_f1 = 0

threshold_results = []

for threshold in thresholds:

    predictions = (
        probabilities >= threshold
    ).astype(int)

    p = precision_score(
        y_test, predictions, zero_division=0
    )

    r = recall_score(
        y_test, predictions, zero_division=0
    )

    f = f1_score(
        y_test, predictions, zero_division=0
    )

    a = accuracy_score(
        y_test, predictions
    )

    threshold_results.append(
        [threshold, a, p, r, f]
    )

    if f > best_f1:
        best_f1 = f
        best_threshold = threshold

# ============================================================
# 19. FINAL OPTIMIZED MODEL RESULTS
# ============================================================

final_predictions = (
    probabilities >= best_threshold
).astype(int)

final_accuracy = accuracy_score(
    y_test, final_predictions
)

final_precision = precision_score(
    y_test, final_predictions, zero_division=0
)

final_recall = recall_score(
    y_test, final_predictions, zero_division=0
)

final_f1 = f1_score(
    y_test, final_predictions, zero_division=0
)

print("\n" + "=" * 60)
print("FINAL RIDENOW AI MODEL")
print("=" * 60)

print(f"Decision Threshold : {best_threshold:.2f}")
print(f"Accuracy           : {final_accuracy*100:.2f}%")
print(f"Precision          : {final_precision*100:.2f}%")
print(f"Recall             : {final_recall*100:.2f}%")
print(f"F1 Score           : {final_f1*100:.2f}%")
print(f"ROC-AUC            : {roc_auc:.4f}")
print(f"PR-AUC             : {pr_auc:.4f}")

# ============================================================
# 20. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    final_predictions
)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Completed", "Cancelled"],
    yticklabels=["Completed", "Cancelled"]
)

plt.title("Ride Cancellation Prediction — Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()
plt.show()

# ============================================================
# 21. CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        final_predictions,
        target_names=["Completed", "Cancelled"],
        zero_division=0
    )
)

# ============================================================
# 22. SAVE FINAL MODEL
# ============================================================

model_package = {
    "model": model,
    "features": list(X.columns),
    "categorical_columns": categorical_cols,
    "threshold": float(best_threshold)
}

joblib.dump(
    model_package,
    "/content/RideNow_cancellation_model.pkl"
)

print("\n" + "=" * 60)
print("✅ RIDENOW AI PIPELINE COMPLETE")
print("=" * 60)

print("Dataset:", df.shape)
print("Model: HistGradientBoostingClassifier")
print("Optimal threshold:", round(best_threshold, 2))
print("Model saved to:")
print("/content/RideNow_cancellation_model.pkl")
```

```python
# =============================================================
# RIDENOW AI — FAST FINAL CANCELLATION MODEL
# Run this AFTER your CSV has been loaded into `df`
# =============================================================

import numpy as np
import pandas as pd
import joblib

from sklearn.model_selection import train_test_split
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    classification_report
)

# Drop rows where the target is NaN before processing
df = df.dropna(subset=["cancelled"]).copy()

print("Dataset:", df.shape)

# -----------------------------
# 1. Separate target
# -----------------------------
target = "cancelled"

X = df.drop(columns=[target, "ride_id"]).copy()
y = df[target].astype(int)

# -----------------------------
# 2. Encode categorical columns
# -----------------------------
categorical_cols = X.select_dtypes(include=["object"]).columns.tolist()

for col in categorical_cols:
    X[col] = X[col].astype("category").cat.codes

# Make everything numeric
X = X.astype(np.float32)

print("Features:", X.shape[1])
print("Cancellation rate:", f"{y.mean()*100:.2f}%")

# -----------------------------
# 3. Train/test split
# -----------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training:", X_train.shape)
print("Testing :", X_test.shape)

# -----------------------------
# 4. Train FAST model
# -----------------------------
print("\nTraining model...")

model = HistGradientBoostingClassifier(
    max_iter=120,
    learning_rate=0.08,
    max_leaf_nodes=31,
    min_samples_leaf=80,
    l2_regularization=1.0,
    early_stopping=True,
    random_state=42
)

model.fit(X_train, y_train)

print("✅ Training complete!")

# -----------------------------
# 5. Probabilities
# -----------------------------
proba = model.predict_proba(X_test)[:, 1]

# First evaluate at standard threshold
threshold = 0.50
pred = (proba >= threshold).astype(int)

# -----------------------------
# 6. Metrics
# -----------------------------
accuracy = accuracy_score(y_test, pred)
precision = precision_score(y_test, pred, zero_division=0)
recall = recall_score(y_test, pred, zero_division=0)
f1 = f1_score(y_test, pred, zero_division=0)
roc_auc = roc_auc_score(y_test, proba)
pr_auc = average_precision_score(y_test, proba)

print("\n" + "="*55)
print("        RIDENOW AI MODEL RESULTS")
print("="*55)

print(f"Accuracy       : {accuracy*100:.2f}%")
print(f"Precision      : {precision*100:.2f}%")
print(f"Recall         : {recall*100:.2f}%")
print(f"F1 Score       : {f1*100:.2f}%")
print(f"ROC-AUC        : {roc_auc:.4f}")
print(f"PR-AUC         : {pr_auc:.4f}")

print("\nClassification Report:")
print(classification_report(
    y_test,
    pred,
    target_names=["Completed", "Cancelled"],
    zero_division=0
))

# -----------------------------
# 7. Find better threshold
# -----------------------------
print("\nFinding best threshold for F1...")

thresholds = np.arange(0.05, 0.51, 0.01)

best_threshold = 0.50
best_f1 = 0

for t in thresholds:
    p = (proba >= t).astype(int)
    score = f1_score(y_test, p, zero_division=0)

    if score > best_f1:
        best_f1 = score
        best_threshold = t

best_pred = (proba >= best_threshold).astype(int)

print("\n" + "="*55)
print("        BEST THRESHOLD RESULTS")
print("="*55)

print(f"Threshold      : {best_threshold:.2f}")
print(f"Accuracy       : {accuracy_score(y_test, best_pred)*100:.2f}%")
print(f"Precision      : {precision_score(y_test, best_pred, zero_division=0)*100:.2f}%")
print(f"Recall         : {recall_score(y_test, best_pred, zero_division=0)*100:.2f}%")
print(f"F1 Score       : {f1_score(y_test, best_pred, zero_division=0)*100:.2f}%")

print("\nClassification Report:")
print(classification_report(
    y_test,
    best_pred,
    target_names=["Completed", "Cancelled"],
    zero_division=0
))

# -----------------------------
# 8. Save model + settings
# -----------------------------
model_package = {
    "model": model,
    "features": list(X.columns),
    "categorical_columns": categorical_cols,
    "threshold": float(best_threshold)
}

joblib.dump(
    model_package,
    "/content/RideNow_cancellation_model.pkl"
)

print("\n✅ MODEL SAVED")
print("/content/RideNow_cancellation_model.pkl")
```
